## 1. A ideia do método

Sem amortecimento e sem forçamento, o pêndulo é conservativo, com hamiltoniana

$$H(\theta, \omega) = \frac{\omega^2}{2} - \cos\theta .$$

O pêndulo invertido, $(\theta, \omega) = (\pm\pi, 0)$, é um ponto de sela com $H = 1$. As órbitas com $H = 1$ que saem da sela em $\theta = -\pi$ e chegam à sela em $\theta = +\pi$ (o mesmo ponto no cilindro) formam a **separatriz**. Sobre ela, a variedade instável da sela ($W^u$, os pontos que se afastam da sela) coincide exatamente com a variedade estável ($W^s$, os pontos que chegam à sela).

Com amortecimento e forçamento, a sela vira uma órbita periódica hiperbólica, e $W^u$ e $W^s$ deixam de coincidir. Há duas possibilidades.

- **As variedades se separam sem se cruzar.** A fronteira entre trajetórias que passam ou não por cima do ponto mais alto é uma curva suave, e o movimento perto da sela é regular.
- **As variedades se cruzam transversalmente.** Como elas são invariantes, um cruzamento (ponto homoclínico) implica infinitos: forma-se um **emaranhado homoclínico**. O teorema de Smale-Birkhoff garante então uma **ferradura de Smale**, um conjunto invariante com órbitas periódicas de todos os períodos e sensibilidade às condições iniciais.

O método de Melnikov decide analiticamente entre as duas possibilidades. Ele calcula, em primeira ordem nas perturbações $q$ e $f$, a distância entre $W^u$ e $W^s$ ao longo da antiga separatriz. Se essa distância troca de sinal quando a fase do forçamento varia, as variedades se cruzam.

## 2. Derivação

### 2.1 Separatriz do pêndulo livre

Em $H = 1$:

$$\frac{\omega^2}{2} - \cos\theta = 1 \;\Longrightarrow\; \omega^2 = 2(1 + \cos\theta) = 4\cos^2\frac{\theta}{2} \;\Longrightarrow\; \omega = 2\cos\frac{\theta}{2} \quad (\text{ramo superior}, \; -\pi < \theta < \pi).$$

Com $u = \theta/2$, a equação $\dot\theta = 2\cos(\theta/2)$ vira $\dot u = \cos u$, e

$$\int \frac{du}{\cos u} = \int dt \;\Longrightarrow\; \ln\left|\sec u + \tan u\right| = t ,$$

escolhendo $\theta(0) = 0$. Logo $\sec u + \tan u = e^{t}$. Tomando $\tan u = \sinh t$, tem-se $\sec u = \sqrt{1+\sinh^2 t} = \cosh t$ e $\sec u + \tan u = \cosh t + \sinh t = e^t$, como exigido. Portanto

$$\theta_0(t) = 2\arctan(\sinh t), \qquad \omega_0(t) = \dot\theta_0 = \frac{2\cosh t}{1 + \sinh^2 t} = 2\,\mathrm{sech}\,t .$$

Em $t = 0$ a órbita passa por $q_0(0) = (0, 2)$, o ponto mais baixo, com velocidade máxima. Duas identidades úteis, de $\sin(2\arctan x) = 2x/(1+x^2)$ e $\cos(2\arctan x) = (1-x^2)/(1+x^2)$:

$$\sin\theta_0 = \frac{2\sinh t}{\cosh^2 t}, \qquad \cos\theta_0 = \frac{1 - \sinh^2 t}{\cosh^2 t}.$$

### 2.2 Função de Melnikov

Escreve-se o sistema da Eq. (3) como uma parte hamiltoniana mais uma perturbação:

$$\dot x = F(x) + g(x, t), \qquad F = \begin{pmatrix} \omega \\ -\sin\theta \end{pmatrix}, \qquad g = \begin{pmatrix} 0 \\ -q\,\omega + f\cos\Omega t \end{pmatrix},$$

com $q$ e $f$ tratados como pequenos. A função de Melnikov é

$$M(t_0) = \int_{-\infty}^{\infty} F\big(q_0(t)\big) \wedge g\big(q_0(t),\, t + t_0\big)\, dt, \qquad F \wedge g = F_1 g_2 - F_2 g_1 .$$

Como $g_1 = 0$, $F \wedge g = \omega_0\,g_2$:

$$M(t_0) = \int_{-\infty}^{\infty} \omega_0(t)\Big[-q\,\omega_0(t) + f\cos\big(\Omega(t + t_0)\big)\Big]dt .$$

**Leitura física.** $\omega\,g_2 = \omega\,\dot\omega\big|_{\text{pert.}}$ é a taxa $dH/dt$ com que a perturbação altera a energia. $M(t_0)$ é, portanto, a variação total de energia acumulada ao longo de uma passagem pela separatriz, que começa em uma fase do forçamento fixada por $t_0$.

**Termo de amortecimento.**

$$-q\int_{-\infty}^{\infty}\omega_0^2\,dt = -4q\int_{-\infty}^{\infty}\mathrm{sech}^2 t\,dt = -4q\Big[\tanh t\Big]_{-\infty}^{\infty} = -4q\,(1 - (-1)) = -8q .$$

**Termo de forçamento.** Com $\cos(\Omega t + \Omega t_0) = \cos\Omega t\cos\Omega t_0 - \sin\Omega t\sin\Omega t_0$:

$$f\int_{-\infty}^{\infty} 2\,\mathrm{sech}\,t\,\cos(\Omega t + \Omega t_0)\,dt = 2f\cos\Omega t_0\underbrace{\int_{-\infty}^{\infty}\mathrm{sech}\,t\cos\Omega t\,dt}_{I(\Omega)} \;-\; 2f\sin\Omega t_0\underbrace{\int_{-\infty}^{\infty}\mathrm{sech}\,t\sin\Omega t\,dt}_{=\,0\ (\text{integrando ímpar})}.$$

### 2.3 A integral $I(\Omega)$ por resíduos

Seja $J = \int_{-\infty}^{\infty} e^{i\Omega t}\,\mathrm{sech}\,t\,dt$; como a parte com seno se anula, $J = I(\Omega)$. Integra-se $e^{i\Omega z}/\cosh z$ no retângulo de vértices $-R$, $R$, $R + i\pi$, $-R + i\pi$, percorrido no sentido anti-horário, com $R \to \infty$.

- **Lado inferior** ($z = x$): contribui $J$.
- **Lado superior** ($z = x + i\pi$, de $R$ para $-R$): $\cosh(x + i\pi) = -\cosh x$ e $e^{i\Omega(x + i\pi)} = e^{-\pi\Omega}e^{i\Omega x}$, logo $\displaystyle\int_R^{-R}\frac{e^{-\pi\Omega}e^{i\Omega x}}{-\cosh x}dx = e^{-\pi\Omega}J$.
- **Lados verticais:** o integrando decai como $e^{-R}$ e as contribuições se anulam.

O único polo interno é o zero de $\cosh z$ em $z = i\pi/2$, com resíduo

$$\mathrm{Res}_{z = i\pi/2}\frac{e^{i\Omega z}}{\cosh z} = \frac{e^{i\Omega(i\pi/2)}}{\sinh(i\pi/2)} = \frac{e^{-\pi\Omega/2}}{i}.$$

Pelo teorema dos resíduos,

$$\left(1 + e^{-\pi\Omega}\right)J = 2\pi i\,\frac{e^{-\pi\Omega/2}}{i} \;\Longrightarrow\; J = \frac{2\pi\,e^{-\pi\Omega/2}}{1 + e^{-\pi\Omega}} = \frac{2\pi}{e^{\pi\Omega/2} + e^{-\pi\Omega/2}} = \pi\,\mathrm{sech}\!\left(\frac{\pi\Omega}{2}\right).$$

### 2.4 Resultado e critério

$$\boxed{\,M(t_0) = -8q + 2\pi f\,\mathrm{sech}\!\left(\frac{\pi\Omega}{2}\right)\cos(\Omega t_0)\,}$$

$M$ oscila em torno de $-8q$ com amplitude $2\pi f\,\mathrm{sech}(\pi\Omega/2)$. Ela tem zeros simples se, e somente se, a amplitude supera o termo constante:

$$2\pi f\,\mathrm{sech}\!\left(\frac{\pi\Omega}{2}\right) > 8q \;\Longleftrightarrow\; \frac{f}{q} > R(\Omega) \equiv \frac{4}{\pi}\cosh\!\left(\frac{\pi\Omega}{2}\right).$$

Nesse caso os zeros satisfazem $\cos\Omega t_0^* = f_c/f$ e são simples, porque $M'(t_0^*) = -2\pi f\Omega\,\mathrm{sech}(\pi\Omega/2)\sin\Omega t_0^* \neq 0$. Em $f = f_c$ os dois zeros se fundem (tangência); para $f < f_c$, $M < 0$ em todas as fases e as variedades não se tocam.

Para os parâmetros do artigo, $\Omega = 2/3$ e $q = 0{,}5$:

$$\cosh\frac{\pi}{3} = 1{,}6003, \qquad R = 2{,}0375, \qquad f_c = q\,R = 1{,}0188 .$$

O ramo inferior da separatriz, $(\theta, \omega) \to (-\theta, -\omega)$, dá $M^-(t_0) = -8q - 2\pi f\,\mathrm{sech}(\pi\Omega/2)\cos\Omega t_0$: o mesmo limiar, com zeros meio período depois, por simetria.

**Distância entre as variedades.** A teoria de Melnikov garante que, na seção de Poincaré do instante $t_0$, a distância entre $W^u$ e $W^s$ medida ao longo da normal à separatriz em $q_0(0) = (0, 2)$ é

$$d(t_0) = \frac{M(t_0)}{\lVert F(q_0(0))\rVert} + O(\varepsilon^2) = \frac{M(t_0)}{2}.$$

Em $(0, 2)$ o campo é $F = (2, 0)$, horizontal, então a normal é a reta $\theta = 0$ e $d = \omega_u - \omega_s$, a diferença entre as alturas em que $W^u$ e $W^s$ cruzam $\theta = 0$. É isso que será medido numericamente.

**Interpretação física.** Em uma passagem pela separatriz, o amortecimento retira a energia $8q$ e o forçamento devolve até $2\pi f\,\mathrm{sech}(\pi\Omega/2)$, conforme a fase. O emaranhado existe quando, em alguma fase, o forçamento consegue repor a energia perdida, de modo que uma trajetória que deixa o pêndulo invertido pode voltar a ele. O fator $\mathrm{sech}(\pi\Omega/2)$ vem da duração finita da passagem ($\omega_0 = 2\,\mathrm{sech}\,t$ é um pulso de largura da ordem de 1): um forçamento rápido demais se cancela durante a passagem, e sua eficiência cai exponencialmente com $\Omega$.

## 3. Verificação simbólica e numérica da derivação

Cada etapa da Seção 2 é conferida: a separatriz satisfaz as equações de movimento e tem $H = 1$; a integral do amortecimento vale 2; o resíduo leva a $\pi\,\mathrm{sech}(\pi\Omega/2)$; e a forma fechada de $M(t_0)$ coincide com a integral de Melnikov calculada por quadratura numérica.

```python
# Dependências: numpy, matplotlib e numba.
# No Colab o Numba normalmente já vem instalado; se faltar, descomente a linha abaixo.
# !pip install numba
# Para esta seção também é preciso o SymPy (já instalado no Colab).

import math
import numpy as np
import sympy as sp
import mpmath as mp

def ok(cond, texto):
    print(("✓ " if cond else "✗ ") + texto)

t = sp.symbols('t', real=True)
Om_s = sp.symbols('Omega', positive=True)

th0 = 2*sp.atan(sp.sinh(t))
w0 = 2/sp.cosh(t)
sin_th0 = sp.expand_trig(sp.sin(th0))      # 2 sinh t/(sinh² t + 1)
cos_th0 = sp.expand_trig(sp.cos(th0))

ok(sp.simplify(sp.diff(th0, t) - w0) == 0,        "dθ0/dt = ω0")
ok(sp.simplify(sp.diff(w0, t) + sin_th0) == 0,    "dω0/dt = -sen θ0")
ok(sp.simplify(w0**2/2 - cos_th0) == 1,           "H(θ0, ω0) = 1 (separatriz)")

I_amort = sp.limit(sp.tanh(t), t, sp.oo) - sp.limit(sp.tanh(t), t, -sp.oo)
ok(I_amort == 2, f"∫ sech² t dt = {I_amort}  ->  termo de amortecimento = -8q")

res = sp.simplify(sp.residue(sp.exp(sp.I*Om_s*t)/sp.cosh(t), t, sp.I*sp.pi/2))
J = 2*sp.pi*sp.I*res/(1 + sp.exp(-sp.pi*Om_s))
ok(sp.simplify((J - sp.pi/sp.cosh(sp.pi*Om_s/2)).rewrite(sp.exp)) == 0,
   f"resíduo em iπ/2 = {res}  ->  I(Ω) = π sech(πΩ/2)")

for Om_v in (0.5, 2/3, 1.0):
    num = mp.quad(lambda x: mp.sech(x)*mp.cos(Om_v*x), [-mp.inf, mp.inf])
    ok(abs(num - mp.pi*mp.sech(mp.pi*Om_v/2)) < 1e-12,
       f"Ω = {Om_v:.4f}: quadratura {float(num):.12f} = π sech(πΩ/2)")

q_v, f_v, Om_v = 0.5, 1.2, 2/3
for t0_v in (0.0, 1.0, 2.5):
    M_quad = mp.quad(lambda s: 2*mp.sech(s)*(-q_v*2*mp.sech(s) + f_v*mp.cos(Om_v*(s + t0_v))), [-mp.inf, 0, mp.inf])
    M_form = -8*q_v + 2*mp.pi*f_v*mp.sech(mp.pi*Om_v/2)*mp.cos(Om_v*t0_v)
    ok(abs(M_quad - M_form) < 1e-10, f"t0 = {t0_v}: M por quadratura = {float(M_quad):+.10f} = forma fechada")

R = 4/math.pi*math.cosh(math.pi*(2/3)/2)
print(f"\ncosh(π/3) = {math.cosh(math.pi/3):.4f}   R(2/3) = {R:.4f}   f_c(q = 0,5) = {0.5*R:.4f}")
```

## 4. O limiar no espaço de parâmetros

A razão crítica $R(\Omega) = (4/\pi)\cosh(\pi\Omega/2)$ divide o plano $(\Omega, f/q)$ em duas regiões. O mínimo é $4/\pi \approx 1{,}27$ em $\Omega \to 0$, e o limiar cresce exponencialmente para forçamentos rápidos. A figura marca o domínio estudado no artigo ($\Omega = 2/3$, $f/q$ de 1,8 a 3,0).

## 5. Cálculo numérico das variedades invariantes

A previsão de Melnikov é perturbativa. Para testá-la, as variedades $W^u$ e $W^s$ são calculadas diretamente do sistema completo, sem aproximação. O procedimento tem quatro etapas.

1. **Órbita de sela.** A órbita periódica hiperbólica que continua o pêndulo invertido é um ponto fixo do mapa estroboscópico $P_{t_0}: x(t_0) \mapsto x(t_0 + T_d)$. Ela é muito instável (o multiplicador instável passa de 600), e o método de Newton aplicado ao mapa de um período só converge a partir de um chute quase perfeito. Usa-se então **tiro múltiplo**: o período é dividido em 10 trechos, cada um com expansão pequena, e o Newton resolve simultaneamente as condições de continuidade entre eles. A solução é continuada em $f$ a partir de $f = 0$, onde a sela é exatamente $(\pi, 0)$.
2. **Direções invariantes.** Os autovetores da jacobiana $DP_{t_0}$ na sela dão as direções instável ($|\mu_u| > 1$) e estável ($|\mu_s| < 1$). Pelo teorema de Liouville, $\mu_u\mu_s = \det DP = e^{-qT_d}$, o que serve de teste.
3. **Domínio fundamental.** Pontos $p + s\,v$ com $s$ entre $h$ e $h\,|\mu|$ (distribuídos em escala logarítmica, $h = 10^{-7}$) formam um segmento de $W^u$ (ou $W^s$) que, iterado pelo mapa (ou pelo mapa inverso, integrando para trás), gera a variedade inteira em ordem, a partir da sela.
4. **Distância.** Segue-se o ramo superior de $W^u$ a partir da sela em $\theta \approx -\pi$ e o ramo superior de $W^s$ a partir da sela em $\theta \approx +\pi$ até o primeiro cruzamento com a reta $\theta = 0$. Cada cruzamento é refinado por bissecção no parâmetro $s$. A distância é $d(t_0) = \omega_u - \omega_s$.

```python
from numba import njit, prange

@njit(cache=True)
def rhs(t, th, w, q, f, Om):
    return w, -q*w - math.sin(th) + f*math.cos(Om*t)

@njit(cache=True)
def rk4(t, th, w, dt, q, f, Om):
    k1x, k1y = rhs(t, th, w, q, f, Om)
    k2x, k2y = rhs(t + dt/2, th + dt*k1x/2, w + dt*k1y/2, q, f, Om)
    k3x, k3y = rhs(t + dt/2, th + dt*k2x/2, w + dt*k2y/2, q, f, Om)
    k4x, k4y = rhs(t + dt, th + dt*k3x, w + dt*k3y, q, f, Om)
    return (th + dt/6*(k1x + 2*k2x + 2*k3x + k4x),
            w + dt/6*(k1y + 2*k2y + 2*k3y + k4y))

@njit(cache=True)
def flow(th, w, t0, nper, spp, q, f, Om, sgn):
    # integra nper períodos a partir do instante t0 (sgn = +1 para frente, -1 para trás)
    T = 2*math.pi/Om
    dt = sgn*T/spp
    t = t0
    for _ in range(nper*spp):
        th, w = rk4(t, th, w, dt, q, f, Om)
        t += dt
    return th, w

@njit(cache=True)
def rhs_var(t, th, w, a, b, c, d, q, f, Om):
    # estado + matriz tangente [[a, c], [b, d]]
    ct = math.cos(th)
    return (w, -q*w - math.sin(th) + f*math.cos(Om*t), b, -ct*a - q*b, d, -ct*c - q*d)

@njit(cache=True)
def seg_jac(th, w, t, nsteps, dt, q, f, Om):
    # fluxo por nsteps passos e sua jacobiana
    a, b, c, d = 1.0, 0.0, 0.0, 1.0
    for _ in range(nsteps):
        k1 = rhs_var(t, th, w, a, b, c, d, q, f, Om)
        k2 = rhs_var(t + dt/2, th + dt*k1[0]/2, w + dt*k1[1]/2, a + dt*k1[2]/2, b + dt*k1[3]/2,
                     c + dt*k1[4]/2, d + dt*k1[5]/2, q, f, Om)
        k3 = rhs_var(t + dt/2, th + dt*k2[0]/2, w + dt*k2[1]/2, a + dt*k2[2]/2, b + dt*k2[3]/2,
                     c + dt*k2[4]/2, d + dt*k2[5]/2, q, f, Om)
        k4 = rhs_var(t + dt, th + dt*k3[0], w + dt*k3[1], a + dt*k3[2], b + dt*k3[3],
                     c + dt*k3[4], d + dt*k3[5], q, f, Om)
        th += dt/6*(k1[0] + 2*k2[0] + 2*k3[0] + k4[0])
        w += dt/6*(k1[1] + 2*k2[1] + 2*k3[1] + k4[1])
        a += dt/6*(k1[2] + 2*k2[2] + 2*k3[2] + k4[2])
        b += dt/6*(k1[3] + 2*k2[3] + 2*k3[3] + k4[3])
        c += dt/6*(k1[4] + 2*k2[4] + 2*k3[4] + k4[4])
        d += dt/6*(k1[5] + 2*k2[5] + 2*k3[5] + k4[5])
        t += dt
    return th, w, np.array([[a, c], [b, d]])

def sela_tiro_multiplo(t0, q, f, Om, spp, m=10, X=None, maxit=50, tol=1e-12):
    # Newton para os m nós da órbita periódica (nó 0 no instante t0)
    T = 2*math.pi/Om
    ns, dt = spp//m, T/spp
    tj = t0 + np.arange(m)*T/m
    if X is None:   # chute: resposta linear em torno de theta = pi
        z = f/complex(-Om**2 - 1.0, q*Om)
        X = np.array([[math.pi + (z*np.exp(1j*Om*tt)).real, (1j*Om*z*np.exp(1j*Om*tt)).real] for tt in tj])
    for _ in range(maxit):
        A = np.zeros((2*m, 2*m))
        G = np.zeros(2*m)
        for j in range(m):
            th1, w1, Jm = seg_jac(X[j, 0], X[j, 1], tj[j], ns, dt, q, f, Om)
            jn = (j + 1) % m
            G[2*j:2*j + 2] = [th1 - X[jn, 0], w1 - X[jn, 1]]
            A[2*j:2*j + 2, 2*j:2*j + 2] = Jm
            A[2*j:2*j + 2, 2*jn:2*jn + 2] -= np.eye(2)
        dX = np.linalg.solve(A, -G).reshape(m, 2)
        X = X + dX
        if np.max(np.abs(dX)) < tol:
            return X, True
    return X, np.max(np.abs(G)) < 1e-9

def sela(t0, q, f, Om, spp):
    # continuação em f a partir de f = 0; devolve ponto, multiplicadores e autovetores
    X = None
    for fk in np.linspace(0.0, f, max(1, int(math.ceil(abs(f)/0.1))) + 1)[1:]:
        X, conv = sela_tiro_multiplo(t0, q, fk, Om, spp, X=X)
        if not conv:
            raise RuntimeError(f"tiro múltiplo não convergiu em f = {fk}")
    x = X[0].copy()
    _, _, Jp = seg_jac(x[0], x[1], t0, spp, 2*math.pi/Om/spp, q, f, Om)
    mu, V = np.linalg.eig(Jp)
    iu, is_ = (0, 1) if abs(mu[0]) > abs(mu[1]) else (1, 0)
    return x, mu[iu].real, V[:, iu].real, mu[is_].real, V[:, is_].real

@njit(parallel=True, cache=True)
def iterar(P0, t0, nper, spp, q, f, Om, sgn):
    # aplica o mapa (sgn=+1) ou o inverso (sgn=-1) nper vezes a cada ponto
    N = P0.shape[0]
    out = np.empty((nper + 1, N, 2))
    T = 2*math.pi/Om
    for i in prange(N):
        th, w = P0[i, 0], P0[i, 1]
        out[0, i, 0] = th
        out[0, i, 1] = w
        t = t0
        for k in range(nper):
            th, w = flow(th, w, t, 1, spp, q, f, Om, sgn)
            t += sgn*T
            out[k + 1, i, 0] = th
            out[k + 1, i, 1] = w
    return out

def primeiro_cruzamento(t0, q, f, Om, spp, tipo, h=1e-7, N=3000, nper=3, s_data=None):
    # omega no primeiro cruzamento do ramo superior de W^u (tipo 'u') ou W^s ('s') com theta = 0
    x, mu_u, vu, mu_s, vs = s_data if s_data is not None else sela(t0, q, f, Om, spp)
    if tipo == 'u':
        p, v = x - np.array([2*math.pi, 0.0]), (vu if vu[0] > 0 else -vu)   # sela em -pi, sai para a direita
        lam, sgn = abs(mu_u), 1
    else:
        p, v = x.copy(), (vs if vs[0] < 0 else -vs)                         # sela em +pi, chega pela esquerda
        lam, sgn = 1/abs(mu_s), -1
    s = h*np.exp(np.linspace(0.0, math.log(lam), N))
    it = iterar(p[None, :] + s[:, None]*v[None, :], t0, nper, spp, q, f, Om, sgn)
    for k in range(nper + 1):
        th = it[k, :, 0]
        idx = np.where((th[:-1] < 0) & (th[1:] >= 0))[0] if tipo == 'u' else np.where((th[:-1] > 0) & (th[1:] <= 0))[0]
        if len(idx):
            a, b = s[idx[0]], s[idx[0] + 1]
            fa = th[idx[0]]
            for _ in range(55):                       # bissecção em s
                m_ = 0.5*(a + b)
                xm = p + m_*v
                thm, wm = flow(xm[0], xm[1], t0, k, spp, q, f, Om, sgn)
                if (thm < 0) == (fa < 0):
                    a, fa = m_, thm
                else:
                    b = m_
            return wm
    return np.nan

def distancia(t0, q, f, Om, spp=200):
    # d(t0) = omega_u - omega_s na reta theta = 0
    sd = sela(t0, q, f, Om, spp)
    return primeiro_cruzamento(t0, q, f, Om, spp, 'u', s_data=sd) - primeiro_cruzamento(t0, q, f, Om, spp, 's', s_data=sd)

def melnikov(t0, q, f, Om):
    return -8*q + 2*np.pi*f/np.cosh(np.pi*Om/2)*np.cos(Om*t0)
```

### 5.1 Testes do procedimento

A sela deve satisfazer $\mu_u\mu_s = e^{-qT_d}$ (Liouville). A distância $d(t_0)$ deve ser independente dos parâmetros numéricos (passos por período, número de pontos do domínio fundamental e deslocamento inicial $h$).

```python
Om = 2/3
T = 2*math.pi/Om
for f in (0.9, 1.2):
    x, mu_u, vu, mu_s, vs = sela(0.0, 0.5, f, Om, 200)
    ok(abs(mu_u*mu_s - math.exp(-0.5*T)) < 1e-8,
       f"f = {f}: sela em t0 = 0: (θ, ω) = ({x[0]:.4f}, {x[1]:.4f}), μu = {mu_u:.1f}, μs = {mu_s:.3e}, μu·μs = {mu_u*mu_s:.6f} = e^(-qT)")

ref = None
for spp_, N_, h_ in ((200, 3000, 1e-7), (400, 3000, 1e-7), (200, 8000, 1e-8), (200, 3000, 1e-6)):
    sd = sela(0.3, 0.5, 1.2, Om, spp_)
    dv = (primeiro_cruzamento(0.3, 0.5, 1.2, Om, spp_, 'u', h=h_, N=N_, s_data=sd)
          - primeiro_cruzamento(0.3, 0.5, 1.2, Om, spp_, 's', h=h_, N=N_, s_data=sd))
    ref = dv if ref is None else ref
    ok(abs(dv - ref) < 1e-6, f"spp = {spp_}, N = {N_}, h = {h_:.0e}:  d(t0 = 0,3) = {dv:+.7f}")
```

## 7. Validação 2: o limiar de tangência

O que decide se há emaranhado é o **máximo** de $d(t_0)$ sobre as fases: as variedades se tocam quando esse máximo passa por zero. A função `dist_max` localiza o máximo por busca em grade e seção áurea em $t_0$, e `f_tangencia` encontra, por falsa posição em $f$, o valor $f_h$ em que $\max_{t_0} d = 0$. Esse é o limiar exato de tangência homoclínica do sistema completo, que se compara com o $f_c$ de Melnikov.

```python
def dist_max(q, f, Om=2/3):
    T = 2*math.pi/Om
    grade = np.linspace(-T/3, T/6, 13)
    vals = np.array([distancia(tt, q, f, Om) for tt in grade])
    i = int(np.nanargmax(vals))
    a, b = grade[max(i - 1, 0)], grade[min(i + 1, len(grade) - 1)]
    g = (math.sqrt(5) - 1)/2
    c, d_ = b - g*(b - a), a + g*(b - a)
    fc_, fd_ = distancia(c, q, f, Om), distancia(d_, q, f, Om)
    for _ in range(18):                      # seção áurea
        if fc_ > fd_:
            b, d_, fd_ = d_, c, fc_
            c = b - g*(b - a)
            fc_ = distancia(c, q, f, Om)
        else:
            a, c, fc_ = c, d_, fd_
            d_ = a + g*(b - a)
            fd_ = distancia(d_, q, f, Om)
    tm = 0.5*(a + b)
    return distancia(tm, q, f, Om), tm

def f_tangencia(q, f_lo, f_hi, Om=2/3):
    g_lo, g_hi = dist_max(q, f_lo, Om)[0], dist_max(q, f_hi, Om)[0]
    for _ in range(40):                      # falsa posição
        fm = f_hi - g_hi*(f_hi - f_lo)/(g_hi - g_lo)
        gm = dist_max(q, fm, Om)[0]
        if gm < 0:
            f_lo, g_lo = fm, gm
        else:
            f_hi, g_hi = fm, gm
        if abs(gm) < 1e-7:
            break
    return fm

fs = np.round(np.arange(0.80, 1.3001, 0.05), 3)
DM = np.array([dist_max(0.5, f) for f in fs])
print(" f     max d numérico   fase do máximo   max M/2 (Melnikov)")
for f, (dm, tm) in zip(fs, DM):
    print(f"{f:.2f}   {dm:+.4f}           Ω t0 = {Om*tm:+.3f}     {-2 + np.pi*f/np.cosh(np.pi/3):+.4f}")

f_c = 0.5*4/np.pi*np.cosh(np.pi/3)
f_h = f_tangencia(0.5, 1.0, 1.1)
print(f"\nLimiar de Melnikov        f_c = {f_c:.4f}")
print(f"Tangência numérica exata  f_h = {f_h:.4f}   (diferença relativa {100*(f_h - f_c)/f_c:+.2f} %)")
```

## 8. As variedades na seção de Poincaré

Para ver o que o critério descreve, os ramos completos de $W^u$ (vermelho) e $W^s$ (azul) são desenhados na seção de Poincaré usada no artigo ($t_0 = 0$). O domínio fundamental é refinado adaptativamente, inserindo pontos onde as imagens consecutivas ficam distantes, para que as curvas saiam contínuas. $W^u$ é iterada 3 períodos para frente e $W^s$ 2 períodos para trás; trechos que o refinamento não resolve são deixados em branco.

```python
def curva_variedade(t0, q, f, Om, spp, tipo, ramo, sd, h=1e-7, N=4000, nper=3, gap=0.02, max_pts=300000):
    x, mu_u, vu, mu_s, vs = sd
    if tipo == 'u':
        v, lam, sgn = (vu if vu[0] > 0 else -vu), abs(mu_u), 1
    else:
        v, lam, sgn = (vs if vs[0] > 0 else -vs), 1/abs(mu_s), -1
    v = ramo*v
    pedacos = []
    logs = np.linspace(0.0, math.log(lam), N)
    for k in range(1, nper + 1):
        L = logs.copy()
        for _ in range(12):                   # refinamento adaptativo
            it = iterar(x[None, :] + (h*np.exp(L))[:, None]*v[None, :], t0, k, spp, q, f, Om, sgn)[-1]
            dist = np.hypot(np.diff(it[:, 0]), np.diff(it[:, 1]))
            ruins = np.where(dist > gap)[0]
            if len(ruins) == 0 or len(L) + len(ruins) > max_pts:
                break
            L = np.sort(np.concatenate([L, 0.5*(L[ruins] + L[ruins + 1])]))
        pedacos.append(it)
    return np.vstack(pedacos)

def para_plot(P, maxstep=0.25):
    th = np.arctan2(np.sin(P[:, 0]), np.cos(P[:, 0]))
    w = P[:, 1].copy()
    salto = np.where((np.abs(np.diff(th)) > math.pi) | (np.hypot(np.diff(th), np.diff(w)) > maxstep))[0]
    return np.insert(th, salto + 1, np.nan), np.insert(w, salto + 1, np.nan)

@njit(cache=True)
def secao(f, q, Om, spp, n_trans, n_rec, th, w):
    T = 2*math.pi/Om
    dt, t = T/spp, 0.0
    for _ in range(n_trans*spp):
        th, w = rk4(t, th, w, dt, q, f, Om)
        t += dt
    TH, W = np.empty(n_rec), np.empty(n_rec)
    for n in range(n_rec):
        for _ in range(spp):
            th, w = rk4(t, th, w, dt, q, f, Om)
            t += dt
        TH[n], W[n] = math.atan2(math.sin(th), math.cos(th)), w
    return TH, W

variedades = {}
for f in (0.9, 1.2):
    sd = sela(0.0, 0.5, f, Om, 200)
    variedades[f] = dict(sela=sd[0],
                         u=[curva_variedade(0.0, 0.5, f, Om, 200, 'u', r, sd, nper=3) for r in (+1, -1)],
                         s=[curva_variedade(0.0, 0.5, f, Om, 200, 's', r, sd, nper=2) for r in (+1, -1)],
                         atrator=secao(f, 0.5, Om, 100, 800, 20000, 0.2, 0.0))
    print(f"f = {f}: variedades calculadas")
```

## 9. Validação 3: consequências dinâmicas do limiar

Se o critério de Melnikov descreve o mecanismo do caos, três coisas devem acontecer em torno de $f_c$:

1. **Nenhum atrator caótico abaixo do limiar.** Sem emaranhado não há ferradura, e o expoente de Lyapunov deve ser negativo para todo $f < f_h$.
2. **As fronteiras de bacia ficam fractais acima do limiar.** No espaço de fases desenrolado ($\theta$ real), cada poço $2\pi k$ tem sua cópia do atrator, e as fronteiras entre as bacias desses poços são formadas por $W^s$ da sela. Abaixo do limiar elas são curvas suaves; acima, o emaranhado as torna fractais (McDonald, Grebogi, Ott e Yorke, 1985; Moon e Li, 1985).
3. **Os descritores lagrangianos enxergam $W^s$.** A crista de $\lVert\nabla L^+\rVert$ do Notebook 06 deve coincidir com a variedade estável calculada aqui.

### 9.1 Expoente de Lyapunov

Mesma varredura do Notebook 03 (continuação em $f$, 400 períodos para $\lambda_{\max}$), agora com $f_c$ e $f_h$ marcados.

```python
@njit(cache=True)
def deriv_tan(tt, x, y, a, b, f):
    return y, -0.5*y - math.sin(x) + f*math.cos(Om*tt), b, -math.cos(x)*a - 0.5*b

@njit(cache=True)
def rk4_tan(t, th, w, dth, dw, dt, f):
    fx1, fy1, fa1, fb1 = deriv_tan(t, th, w, dth, dw, f)
    t2 = t + dt/2
    fx2, fy2, fa2, fb2 = deriv_tan(t2, th + dt*fx1/2, w + dt*fy1/2, dth + dt*fa1/2, dw + dt*fb1/2, f)
    fx3, fy3, fa3, fb3 = deriv_tan(t2, th + dt*fx2/2, w + dt*fy2/2, dth + dt*fa2/2, dw + dt*fb2/2, f)
    fx4, fy4, fa4, fb4 = deriv_tan(t + dt, th + dt*fx3, w + dt*fy3, dth + dt*fa3, dw + dt*fb3, f)
    return (th + (dt/6)*(fx1 + 2*fx2 + 2*fx3 + fx4), w + (dt/6)*(fy1 + 2*fy2 + 2*fy3 + fy4),
            dth + (dt/6)*(fa1 + 2*fa2 + 2*fa3 + fa4), dw + (dt/6)*(fb1 + 2*fb2 + 2*fb3 + fb4))

@njit(cache=True)
def varredura_lyap(f_arr, spp, n_trans, n_lyap):
    T = 2*math.pi/Om
    dt = T/spp
    th, w = 0.2, 0.0
    lam = np.empty(len(f_arr))
    for i in range(len(f_arr)):
        f = f_arr[i]
        t = 0.0
        for n in range(n_trans*spp):
            th, w, _, _ = rk4_tan(t, th, w, 0.0, 0.0, dt, f)
            t += dt
        thl, wl, a, b, s, t2 = th, w, 1.0, 0.0, 0.0, 0.0
        for n in range(n_lyap):
            for _ in range(spp):
                thl, wl, a, b = rk4_tan(t2, thl, wl, a, b, dt, f)
                t2 += dt
            r = math.hypot(a, b)
            s += math.log(r)
            a /= r
            b /= r
        lam[i] = s/(n_lyap*T)
    return lam

f_arr = np.linspace(0.9, 1.5, 400)
lam = varredura_lyap(f_arr, 80, 350, 400)
print(f"maior λ_max para f < f_h:  {lam[f_arr < f_h].max():+.4f}")
print(f"primeiro f com λ_max > 0:  {f_arr[lam > 0][0]:.4f}   ({f_arr[lam > 0][0] - f_h:+.4f} acima de f_h)")
```

### 9.2 Fronteiras de bacia: expoente de incerteza

Para $N$ condições iniciais sorteadas em $\theta_0 \in [-\pi, \pi]$, $\omega_0 \in [-2{,}5;\ 2{,}5]$, integra-se 150 períodos e registra-se o estado final **desenrolado** (qual atrator e em qual poço $2\pi k$). Repete-se com $\theta_0 + \epsilon$. A condição inicial é **incerta** se os dois estados finais diferem. A fração de condições incertas escala como

$$\phi(\epsilon) \propto \epsilon^{\alpha}, \qquad D_{\text{fronteira}} = 2 - \alpha .$$

Fronteira suave: $\alpha = 1$ ($D = 1$). Fronteira fractal: $\alpha < 1$.

O estado final desenrolado separa dois tipos de incerteza, e a distinção é essencial.

- **Mesmo atrator, poço diferente.** A condição inicial e sua vizinha passam um número diferente de vezes por cima do ponto mais alto. A fronteira entre poços é formada por $W^s$ do pêndulo invertido, que é exatamente o objeto do critério de Melnikov.
- **Atrator diferente.** Perto de $f \approx 1{,}005$ ocorre uma quebra de simetria: o atrator periódico único se divide em dois atratores assimétricos. A fronteira entre eles é organizada pela órbita simétrica que perdeu estabilidade, outra sela, que não tem órbita homoclínica no pêndulo livre e está fora do alcance deste cálculo de Melnikov.

Só entram as condições iniciais que se acomodaram em uma órbita de período 1 (critério $\lVert x_{n+1} - x_n\rVert < 10^{-3}$). O ajuste usa as escalas $\epsilon$ com pelo menos 10 condições incertas do tipo considerado.

```python
@njit(parallel=True, cache=True)
def estado_final(TH0, W0, q, f, spp, nper):
    N = TH0.size
    out = np.empty((N, 3))
    T = 2*math.pi/Om
    dt = T/spp
    for i in prange(N):
        th, w, t = TH0[i], W0[i], 0.0
        for _ in range(nper*spp):
            th, w = rk4(t, th, w, dt, q, f, Om)
            t += dt
        th1, w1 = th, w
        for _ in range(spp):
            th1, w1 = rk4(t, th1, w1, dt, q, f, Om)
            t += dt
        out[i, 0], out[i, 1], out[i, 2] = th, w, max(abs(th1 - th), abs(w1 - w))
    return out

def ajuste_alpha(epss, cont, n):
    m = cont >= 10
    if m.sum() < 3:
        return np.nan, np.nan
    x, y = np.log(epss[m]), np.log(cont[m]/n[m])
    p = np.polyfit(x, y, 1)
    se = math.sqrt(np.sum((y - np.polyval(p, x))**2)/max(m.sum() - 2, 1)/np.sum((x - x.mean())**2))
    return p[0], se

def expoente_incerteza(f, N=20000, nper=150, seed=3,
                       epss=(1e-1, 3e-2, 1e-2, 3e-3, 1e-3, 3e-4, 1e-4, 3e-5, 1e-5)):
    rng = np.random.default_rng(seed)
    TH0, W0 = rng.uniform(-math.pi, math.pi, N), rng.uniform(-2.5, 2.5, N)
    A = estado_final(TH0, W0, 0.5, f, 100, nper)
    acomodado = A[:, 2] < 1e-3
    c_poco, c_atr, n_bom = [], [], []
    for e in epss:
        B = estado_final(TH0 + e, W0, 0.5, f, 100, nper)
        bom = acomodado & (B[:, 2] < 1e-3)
        dth, dw = B[:, 0] - A[:, 0], np.abs(B[:, 1] - A[:, 1])
        k = np.round(dth/(2*math.pi))
        mesmo_atrator = (np.abs(dth - 2*math.pi*k) < 0.1) & (dw < 0.1)
        c_poco.append(np.sum(bom & mesmo_atrator & (k != 0)))    # outro poço
        c_atr.append(np.sum(bom & ~mesmo_atrator))               # outro atrator
        n_bom.append(bom.sum())
    epss, c_poco, c_atr, n_bom = map(np.array, (epss, c_poco, c_atr, n_bom))
    return ajuste_alpha(epss, c_poco, n_bom), ajuste_alpha(epss, c_atr, n_bom), c_poco, c_atr

f_bacias = [0.95, 1.00, 1.01, 1.02, 1.04, 1.06]
bacias = {}
print("                 fronteira entre POÇOS          fronteira entre ATRATORES")
for f in f_bacias:
    (ap, sp_), (aa, sa), cp, ca = expoente_incerteza(f)
    bacias[f] = (ap, sp_, aa, sa)
    lado = "<" if f < f_h else ">"
    txt_a = f"α = {aa:.2f} ± {sa:.2f}" if not np.isnan(aa) else "um só atrator  "
    print(f"f = {f:.2f} ({lado} f_h):  α = {ap:.2f} ± {sp_:.2f} (D = {2 - ap:.2f})     {txt_a}")
    print(f"      contagens por ε, poço: {cp.tolist()}   atrator: {ca.tolist()}")
```

### 9.3 Descritores lagrangianos e a variedade estável

$\lVert\nabla L^+\rVert$ com $\tau = 3T_d$ e $t_0 = 0$, como na Figura 6 do artigo, com a variedade estável $W^s$ da sela sobreposta em azul.

```python
@njit(parallel=True, cache=True)
def grade_ld(theta_vals, omega_vals, tau, dt, q, f):
    N_om, N_th = len(omega_vals), len(theta_vals)
    ld = np.zeros((N_om, N_th))
    steps = int(tau/dt)
    for i in prange(N_om):
        for j in range(N_th):
            th, om, t, L = theta_vals[j], omega_vals[i], 0.0, 0.0
            for _ in range(steps):
                k1 = rhs(t, th, om, q, f, Om)
                s1 = math.sqrt(k1[0]**2 + k1[1]**2)
                k2 = rhs(t + dt/2, th + dt/2*k1[0], om + dt/2*k1[1], q, f, Om)
                s2 = math.sqrt(k2[0]**2 + k2[1]**2)
                k3 = rhs(t + dt/2, th + dt/2*k2[0], om + dt/2*k2[1], q, f, Om)
                s3 = math.sqrt(k3[0]**2 + k3[1]**2)
                k4 = rhs(t + dt, th + dt*k3[0], om + dt*k3[1], q, f, Om)
                s4 = math.sqrt(k4[0]**2 + k4[1]**2)
                th += dt/6*(k1[0] + 2*k2[0] + 2*k3[0] + k4[0])
                om += dt/6*(k1[1] + 2*k2[1] + 2*k3[1] + k4[1])
                L += dt/6*(s1 + 2*s2 + 2*s3 + s4)
                t += dt
            ld[i, j] = L
    return ld

from matplotlib.colors import LogNorm
th_v, om_v = np.linspace(-math.pi, math.pi, 300), np.linspace(-0.6, 2.5, 300)
fig, axes = plt.subplots(1, 2, figsize=(11, 4.4))
for ax, f in zip(axes, (0.9, 1.2)):
    ld = grade_ld(th_v, om_v, 3*T, T/100, 0.5, f)
    dL_dom, dL_dth = np.gradient(ld, om_v, th_v)
    g = np.sqrt(dL_dth**2 + dL_dom**2)
    ax.pcolormesh(th_v, om_v, g, shading='auto', cmap='inferno',
                  norm=LogNorm(vmin=max(np.percentile(g, 5), 1e-3), vmax=np.percentile(g, 99.5)))
    for P in variedades[f]['s']:
        th_p, w_p = para_plot(P)
        ax.plot(th_p, w_p, color='#4fc3f7', lw=0.9, alpha=0.9)
    ax.set_xlim(-math.pi, math.pi); ax.set_ylim(-0.6, 2.5)
    ax.set_xticks([-math.pi, -math.pi/2, 0, math.pi/2, math.pi], [r'$-\pi$', r'$-\pi/2$', '0', r'$\pi/2$', r'$\pi$'])
    ax.set_xlabel(r'$\theta_0$'); ax.set_ylabel(r'$\omega_0$')
    ax.set_title(fr'$\|\nabla L^+\|$ ($\tau=3T_d$) e $W^s$ (azul), $f={f}$'.replace('.', '{,}'), fontsize=10)
fig.tight_layout()
fig.savefig(FIG_DIR / 'melnikov_ld_variedade_estavel.png', dpi=200)
plt.show()
```

## 10. A ferradura de Smale na seção de Poincaré

O teorema de Smale-Birkhoff garante que, quando as variedades invariantes se entrelaçam transversalmente (acima de $f_c$), existe uma estrutura invariante com caos topológico completo: a **ferradura de Smale**. Ela se manifesta na seção de Poincaré estroboscópica como uma estrutura que se estende, dobra e volta sobre si mesma, revelando como a variedade instável $W^u$ é esticada e reenrolada no espaço de fases.

Aqui visualizamos essa estrutura integrando condições iniciais sobre $W^u$ e seguindo sua evolução através de várias seções de Poincaré sucessivas.

**Validação 3: as consequências dinâmicas.**

- *Lyapunov.* O maior $\lambda_{\max}$ abaixo de $f_h$ é $-0{,}0011$, atingido na quebra de simetria em $f \approx 1{,}005$, onde um multiplicador de Floquet toca $+1$ sem que o expoente fique positivo. O primeiro $\lambda_{\max} > 0$ está em $f = 1{,}0835$, 0,068 acima de $f_h$. Não há atrator caótico abaixo do limiar.
- *Bacias.* A fronteira entre poços, governada pela sela do pêndulo invertido, é suave ($\alpha$ entre 0,93 e 1,13, compatível com 1) até $f = 1{,}01$ e fractal a partir de $f = 1{,}02$ ($\alpha = 0{,}72$, $D = 1{,}28$), com dimensão crescente: $D = 1{,}62$ em $f = 1{,}04$ e $1{,}78$ em $f = 1{,}06$. A transição ocorre no intervalo que contém $f_h$ e $f_c$.
- *Descritores lagrangianos.* A crista única de $\lVert\nabla L^+\rVert$ em $f = 0{,}9$ coincide com $W^s$, e em $f = 1{,}2$ as cristas dobradas seguem as dobras de $W^s$.

**Ressalvas.** O critério é de primeira ordem, e a concordância de 0,34 % é um resultado numérico destes parâmetros, não uma garantia geral. Ele assegura uma ferradura, que pode ser não atratora: entre $f_h$ e $f = 1{,}0835$ o atrator é periódico, e o caos aparece como sensibilidade do estado final (fronteiras fractais). E ele descreve a sela do pêndulo invertido: a fronteira entre os dois atratores assimétricos criados em $f \approx 1{,}005$ já é fractal em $f = 1{,}01$, porque é organizada por outra sela, fora do alcance deste cálculo.

**Conclusão.** O método de Melnikov dá, sem integrar nenhuma trajetória, o limiar $f_c = (4q/\pi)\cosh(\pi\Omega/2)$ e o mecanismo do caos, o emaranhado homoclínico da sela do pêndulo invertido. A previsão coincide com o limiar exato a 0,34 % e é confirmada por três diagnósticos independentes. É, portanto, uma ferramenta analítica e preditiva para a análise do caos, complementar ao expoente de Lyapunov: este diz se uma trajetória é caótica; o método de Melnikov diz onde, no espaço de parâmetros, o caos se torna possível e por quê.

### 11.1 Sobre a ferradura de Smale

A visualização acima mostra como a variedade instável $W^u$ da sela evolui sob o mapa de Poincaré em tempo estroboscópico. Em cada seção, vê-se a variedade esticada pela dinâmica hiperbólica (o multiplicador $\mu_u \gg 1$), formando uma estrutura alongada. Ao retornar à próxima seção de Poincaré, ela volteia sobre si mesma — este é o mecanismo da ferradura.

**Características visíveis:**

1. **Alongamento exponencial:** a largura da estrutura em cada seção cai como $\mu_u^{-k}$, enquanto o comprimento cresce como $\mu_u^k$. Dados os multiplicadores acima, a estrutura fica fino e comprido após poucos períodos.

2. **Múltiplas camadas:** o retorno ao espaço $(\theta, \omega)$ não é único — a variedade passa várias vezes por cada região, criando a multiplicidade característica da ferradura.

3. **Cascata de bifurcações:** cada novo "passe" da variedade sobre si mesma adiciona uma órbita periódica de período dobrado (bifurcação de duplicação de período), levando ao infinito topológico de órbitas periódicas dentro do atrator caótico.

O método de Melnikov previu exatamente em que valor de $f$ essa dinâmica começa a ser possível ($f = f_c$), e o cálculo numérico das variedades confirmou a predição. Esta seção visualiza a consequência geométrica da previsão: o emaranhado homoclínico e a ferradura de Smale.
